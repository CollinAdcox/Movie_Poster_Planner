   Can we find if the style of movie poster impacts how much a movie earns and how critics score it?

```python
##Bringing in API key and AI Model
from google import genai
from google.colab import userdata
key = userdata.get("GOOGLE_API_KEY")
client = genai.Client(api_key=key)
MODEL = "gemini-flash-lite-latest"
```

# Preparing Data for Prompting and Initial Tests


```python
Styles = ["Ensemble Collage", "Main Character Close up", "Orange and Teal Contrast", "Minimalist", "Tiny Subject, Massive Scale", "Character Pairing", "Heavy Typography", "Protagonist from the back", "Two Tone Palette", "Dynamic Motion"]

def clean_label(text, allowed):
    label= text.strip().lower().strip(".")
    if label in allowed:
      return label
    else:
      return "Unsure"
```

```python
def make_prompt(description):
    prompt = f"is this movie poster described as being Ensemble Collage, Main Character Close up, Orange and Teal Contrast, Minimalist, Tiny Subject, Massive Scale, Character Pairing, Heavy Typography, Protagonist from the back, Two Tone Palette, or Dynamic Motion. Answer only with one of those categories. \n\nDescription: {description}"
    return prompt

print(make_prompt(Movies["Poster Description"][4]))
```

```python
def get_style(description):
  import time
  time.sleep(4)
  prompt = make_prompt(description)
  response = client.models.generate_content (
      model=MODEL, contents=prompt)
  return response.text.strip().lower()

my_function = get_style
my_result = get_style(Movies["Poster Description"][4])
print(my_result)
```
